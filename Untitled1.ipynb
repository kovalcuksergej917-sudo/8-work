```python
def validate_price() -> float:
    """Запрашивает и валидирует цену товара (>0 и ≤ 1 000 000)."""
    while True:
        raw_input = input("Введите цену товара: ").strip()
        price_str = raw_input.replace(',', '.')

        try:
            price = float(price_str)
            if price <= 0:
                print("Ошибка: Цена должна быть положительной.\n")
            elif price > 1_000_000:
                print("Ошибка: Цена не должна превышать 1 000 000.\n")
            else:
                return price
        except ValueError:
            print("Ошибка: Введите корректное число (например, 150 или 99.99).\n")

if __name__ == "__main__":
    print(f"✅ Цена принята: {validate_price():.2f} руб.")
```
